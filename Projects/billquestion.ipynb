```python
hotel_name = "Welcome to Sakshi's Eatery"

#Menu
menu = {
    1: ("idli", 30),
    2: ("dosa", 50),
    3: ("poha", 25),
    4: ("upma", 20)
}

#Show menu
print("--------- MENU ---------")
for sr, (item, price) in menu.items():
    print(f"{sr}. {item} - ₹{price}")
print("------------------------")

#User selects item
choice = None
while choice is None:
    try:
        choice_input = input("Enter the sr number of the item you want: ")
        choice = int(choice_input)
        if choice not in menu:
            print("Invalid serial number. Please choose from the menu.")
            choice = None
    except ValueError:
        print("Invalid input. Please enter a number corresponding to the item.")
        choice = None

quantity = None
while quantity is None:
    try:
        quantity_input = input("Enter quantity: ")
        quantity = int(quantity_input)
        if quantity <= 0:
            print("Quantity must be a positive number.")
            quantity = None
    except ValueError:
        print("Invalid quantity. Please enter a number.")
        quantity = None


#Get item and price
item_name, item_price = menu[choice]

#items list for bill
items = [(choice, item_name, quantity, item_price)]

total = sum(qty * price for _, _, qty, price in items)

print("=" * 50)
print(f"|{hotel_name.center(48)}|")
print("=" * 50)
print(f"| {'sr':<3} {'Menu':<15} {'qunt':<8} {'price':>17} |")
print("-" * 50)

for sr, menu_item, qty, price in items:
    print(f"| {sr:<3} {menu_item:<15} {qty:<8} {price:>17} |")

print("-" * 50)
print(f"| {'Total':<30} {total:>15} |")
print("=" * 50)
```

```python
#using oops concept

class MenuItem:
    def __init__(self, sr, name, price):
        self.sr = sr
        self.name = name
        self.price = price


class Menu:
    def __init__(self):
        self.items = {
            1: MenuItem(1, "idli", 30),
            2: MenuItem(2, "dosa", 50),
            3: MenuItem(3, "poha", 25),
            4: MenuItem(4, "upma", 20)
        }

    def display_menu(self):
        print("--------- MENU ---------")
        for item in self.items.values():
            print(f"{item.sr}. {item.name} - ₹{item.price}")
        print("------------------------")

    def get_item(self, sr):
        return self.items.get(sr, None)


class Order:
    def __init__(self):
        self.selected_items = []

    def add_item(self, item, quantity):
        self.selected_items.append((item, quantity))

    def calculate_total(self):
        return sum(item.price * qty for item, qty in self.selected_items)


class BillPrinter:
    def __init__(self, hotel_name):
        self.hotel_name = hotel_name

    def print_bill(self, order):
        print("=" * 50)
        print(f"|{self.hotel_name.center(48)}|")
        print("=" * 50)
        print(f"| {'sr':<3} {'Menu':<15} {'qunt':<8} {'price':>17} |")
        print("-" * 50)

        for item, qty in order.selected_items:
            print(f"| {item.sr:<3} {item.name:<15} {qty:<8} {item.price:>17} |")

        print("-" * 50)
        total = order.calculate_total()
        print(f"| {'Total':<30} {total:>15} |")
        print("=" * 50)

hotel_name = "Welcome to Sakshi's Eatery"
menu = Menu()
order = Order()
bill = BillPrinter(hotel_name)

menu.display_menu()

#Select item
while True:
    try:
        choice = int(input("Enter the sr number of the item you want: "))
        selected_item = menu.get_item(choice)
        if selected_item:
            break
        else:
            print("Invalid serial number. Try again.")
    except ValueError:
        print("Enter a valid number.")

#Quantity
while True:
    try:
        quantity = int(input("Enter quantity: "))
        if quantity > 0:
            break
        else:
            print("Quantity must be positive.")
    except ValueError:
        print("Enter a valid number.")

#Add to order
order.add_item(selected_item, quantity)

bill.print_bill(order)
```

```python
import sqlite3

#Create database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

#Create menu table
cursor.execute("""
CREATE TABLE menu(
    sr INTEGER PRIMARY KEY,
    item TEXT,
    price INTEGER
);
""")

#Inserting menu items
cursor.executemany("""
INSERT INTO menu VALUES (?, ?, ?);
""", [
    (1, "idli", 30),
    (2, "dosa", 50),
    (3, "poha", 25),
    (4, "upma", 20)
])

conn.commit()
```

```python
print("--------- MENU ---------")
result = cursor.execute("SELECT * FROM menu").fetchall()

for row in result:
    print(f"{row[0]}. {row[1]} - ₹{row[2]}")
print("------------------------")
```

```python
query = """
SELECT
    m.sr,
    m.item,
    ? AS quantity,
    m.price,
    (m.price * ?) AS total_price
FROM menu m
WHERE m.sr = ?;
"""

bill_item = cursor.execute(query, (quantity, quantity, choice)).fetchone()
```
