```python
#12. Ask user to enter information about salary of the employee per year and rating received as A, B, C, D

#If salary upto 5 lpa then increament based on ratings are A = 16% , B=12%, C= 10%, D=6%

#If salary upto 10 lpa then increament based on ratings are A = 14% , B=10%, C= 8%, D=6%

#If salary upto 15 lpa then increament based on ratings are A = 8% , B=6%, C= 4%, D = no increment

#If salary upto 23 lpa then increament based on ratings are A = 7% , B=5%, C= 4%, D=no
```

```python
salary = float(input("Enter your salary per year(in LPA): "))
rating = input("Enter your rating (A/B/C/D):").upper()

#increment percentages for each salary slab
if salary <= 5:
    increments = {'A': 16, 'B': 12, 'C': 10, 'D': 6}
elif salary <= 10:
    increments = {'A': 14, 'B': 10, 'C': 8, 'D': 6}
elif salary <= 15:
    increments = {'A': 8, 'B': 6, 'C': 4, 'D': 0}
elif salary <= 23:
    increments = {'A': 7, 'B': 5, 'C': 4, 'D': 0}
else:
    increments = {'A': 0, 'B': 0, 'C': 0, 'D': 0}

# Get increment percentage from the dictionary
increment = increments.get(rating, 0)

# Calculate new salary
new_salary = salary + (salary * increment / 100)

# Print result
print(f"Original Salary: {salary} LPA")
print(f"Rating: {rating}")
print(f"Increment: {increment}%")
print(f"New Salary: {new_salary} LPA")
```

13. Ask user to opt for courses for master degree based on the following

L1 = [“HR”, “Finance”, “Marketing”, “DS”]

Based on above subject there are two different streams. For example- HR is having HR core and HR analytics and Marketing is having core and Marketing analytics. Analytics is the optional subject and having added extra fees. DS is not having analytics.

If fees for L1 is 2 lakhs for each course core subject having the same fees but analytics subject having 10% extra on 2 lakhs.

If student opts for hostel then 2 lakhs per year is added. For food monthly 2000 .

Transportation charges 13000 per semester. Calculate the total annual cost based on selected service.

User will enter values as subject, analytics(Y/N), Hostel (Y/N), food(How many months?), Transportation(semester/annual)

```python
#using if else
L1 = ["HR", "Finance", "Marketing", "DS"]

#base fees
BASE_FEE = 200000            # 2 lakhs
ANALYTICS_EXTRA = 0.10       # 10%
HOSTEL_FEE = 200000          # per year
FOOD_PER_MONTH = 2000
TRANSPORT_PER_SEM = 13000

#user inputs
subject = input("Enter subject (HR/Finance/Marketing/DS):").strip()
analytics = input("Analytics? (Y/N):").strip().upper()
hostel = input("Hostel? (Y/N): ").strip().upper()
food_months = int(input("Food for how many months?:"))
transport = input("Transportation (semester/annual):").strip().lower()

if subject not in L1:
    print("Invalid subject selected")
    exit()

total_fee = BASE_FEE

if analytics == "Y" and subject != "DS":
    total_fee += BASE_FEE * ANALYTICS_EXTRA

if hostel == "Y":
    total_fee += HOSTEL_FEE

total_fee += food_months * FOOD_PER_MONTH

if transport == "semester":
    total_fee += TRANSPORT_PER_SEM * 2   #2 semesters per year
elif transport == "annual":
    total_fee += TRANSPORT_PER_SEM * 2
else:
    print("Invalid transportation option")
    exit()

print("\n----- Fee Breakdown ----")
print(f"Subject: {subject}")
print(f"Total Annual Cost in Rs:{int(total_fee)}")
```

```python
import sqlite3

#SQLite DB
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

#Create tables
cursor.execute("""
CREATE TABLE fee_structure (
    subject TEXT PRIMARY KEY,
    base_fee INTEGER,
    analytics_extra REAL
);
""")

cursor.execute("""
CREATE TABLE hostel_food_transport (
    hostel_fee INTEGER,
    food_per_month INTEGER,
    transport_per_sem INTEGER
);
""")

#Insert fee structure
cursor.executemany("""
INSERT INTO fee_structure VALUES (?, ?, ?);
""", [
    ("HR",      200000, 0.10),
    ("Finance", 200000, 0.10),
    ("Marketing", 200000, 0.10),
    ("DS",      200000, 0.00)
])

cursor.execute("""
INSERT INTO hostel_food_transport VALUES (200000, 2000, 13000);
""")

conn.commit()
```

```python
query = """
SELECT
    f.subject,
    f.base_fee
    + CASE
        WHEN ? = 'Y' AND f.subject != 'DS'
        THEN f.base_fee * f.analytics_extra
        ELSE 0
      END
    + CASE
        WHEN ? = 'Y' THEN h.hostel_fee ELSE 0
      END
    + (? * h.food_per_month)
    + CASE
        WHEN ? = 'semester' THEN h.transport_per_sem * 2
        WHEN ? = 'annual' THEN h.transport_per_sem * 2
        ELSE 0
      END
      AS total_fee
FROM fee_structure f, hostel_food_transport h
WHERE f.subject = ?;
"""

result = cursor.execute(
    query,
    (analytics, hostel, food_months, transport, transport, subject)
).fetchone()

print("\n----- Fee Breakdown -----")
print(f"Subject: {result[0]}")
print(f"Total Annual Cost: ₹{int(result[1])}")
```

```python
#using oops concept

class CourseFee:

    #Class Variables
    SUBJECTS = ["HR", "Finance", "Marketing", "DS"]

    BASE_FEE = 200000
    ANALYTICS_EXTRA = 0.10
    HOSTEL_FEE = 200000
    FOOD_PER_MONTH = 2000
    TRANSPORT_PER_SEM = 13000

    def __init__(self):
        self.subject = ""
        self.analytics = ""
        self.hostel = ""
        self.food_months = 0
        self.transport = ""
        self.total_fee = 0

    def take_input(self):

        while True:
            subject = input("Enter subject (HR/Finance/Marketing/DS):").strip()
            if subject in self.SUBJECTS:
                self.subject = subject
                break
            print("Invalid subject.Choose from HR, Finance, Marketing, DS.")

        while True:
            analytics = input("Analytics? (Y/N):").strip().upper()
            if analytics in ["Y", "N"]:
                self.analytics = analytics
                break
            print("Enter Y or N only.")

        while True:
            hostel = input("Hostel? (Y/N):").strip().upper()
            if hostel in ["Y", "N"]:
                self.hostel = hostel
                break
            print("Enter Y or N only")

        while True:
            try:
                food_months = int(input("Food for how many months:"))
                if food_months >= 0:
                    self.food_months = food_months
                    break
                print("Months cannot be negative.")
            except ValueError:
                print("Please enter a valid number.")

        while True:
            transport = input("Transportation (semester/annual):").strip().lower()
            if transport in ["semester", "annual"]:
                self.transport = transport
                break
            print("Enter 'semester' or 'annual' only.")

    #Calculations
    def calculate_total_fee(self):

        self.total_fee = self.BASE_FEE

        if self.analytics == "Y" and self.subject != "DS":
            self.total_fee += self.BASE_FEE * self.ANALYTICS_EXTRA

        if self.hostel == "Y":
            self.total_fee += self.HOSTEL_FEE

        self.total_fee += self.food_months * self.FOOD_PER_MONTH

        #Annual means 2 semesters
        if self.transport == "annual":
            self.total_fee += self.TRANSPORT_PER_SEM * 2
        else:
            self.total_fee += self.TRANSPORT_PER_SEM

    def display_fee(self):
        print("\n----- Fee Breakdown ----")
        print(f"Subject: {self.subject}")
        print(f"Total Annual Cost in Rs: {int(self.total_fee)}")



if __name__ == "__main__":
    obj = CourseFee()
    obj.take_input()
    obj.calculate_total_fee()
    obj.display_fee()
```

14. Digitalize the book allotment process for school. Charges are mentioned here in the given table:

```python

#data
books_charges = {
    "Hindi": {"1-4": 60, "5-8": 100, "9-10": 150},
    "Marathi": {"1-4": 60, "5-8": 100, "9-10": 150},
    "English": {"1-4": 80, "5-8": 100, "9-10": 150},
    "Science": {"1-4": 90, "5-8": 120, "9-10": 200},
    "Maths": {"1-4": 100, "5-8": 140, "9-10": 250}
}

notebook_charges = {
    "square": {"100": 40, "200": 70},
    "4lines": {"100": 30, "200": 50},
    "2lines": {"100": 30, "200": 50},
    "single lines": {"100": 60, "200": 100},
    "a4 notebook": {"100": 100, "200": 180}
}

standard = int(input("Enter standard (1-10): "))

if 1 <= standard <= 4:
    std_group = "1-4"
elif 5 <= standard <= 8:
    std_group = "5-8"
elif 9 <= standard <= 10:
    std_group = "9-10"
else:
    print("Invalid standard")
    exit()

subjects = input(
    "Enter subjects (comma separated): "
).split(",")

notebook_type = input(
    "Enter notebook type (square / 4lines / 2lines / single lines / a4 notebook): "
).lower()

pages = input("Enter pages (100 / 200): ")

total_amount = 0

print("\n========== BOOK ALLOTMENT BILL ==========")
print(f"Standard : {standard}")
print("\nBooks:")

for subject in subjects:
    subject = subject.strip().title()
    if subject in books_charges:
        price = books_charges[subject][std_group]
        total_amount += price
        print(f"{subject:<10} : ₹{price}")
    else:
        print(f"{subject:<10} : Not Available")

print("\nNotebook:")

if notebook_type in notebook_charges and pages in notebook_charges[notebook_type]:
    nb_price = notebook_charges[notebook_type][pages]
    total_amount += nb_price
    print(f"{notebook_type.title()} ({pages} pages) : ₹{nb_price}")
else:
    print("Invalid notebook selection")

print("-----------------------------------------")
print(f"TOTAL AMOUNT : ₹{total_amount}")
print("-----------------------------------------")
```

```python
restaurant = {}

menu = {
    1: ("Pizza", 200),
    2: ("Burger", 120),
    3: ("Coke", 40),
    4: ("Pasta", 150),
    5: ("Coffee", 80),
}

def take_order():
    table_no = input("Enter table number: ")

    print("\n--- MENU ---")
    for num, (item, price) in menu.items():
        print(f"{num}. {item} - ₹{price}")

    print("\nSelect items (comma separated numbers or names):")
    choices = input("Your order: ").split(",")

    items = []          #list of item names
    quantities = []     #list of quantities

    for user_choice in choices:
        choice = user_choice.strip().lower()


        if choice.isdigit() and int(choice) in menu:
            item_name = menu[int(choice)][0]
        else:

            item_name = None
            for num, (name, _) in menu.items():
                if choice == name.lower():
                    item_name = name
                    break

        if item_name is None:
            print(f"Invalid menu option: {user_choice}")
            continue

        qty = input(f"Enter quantity for {item_name}: ")
        if qty.isdigit():
            qty = int(qty)
        else:
            print("Invalid quantity. Setting quantity = 1")
            qty = 1

        items.append(item_name)
        quantities.append(qty)

    restaurant[table_no] = {
        "order": items,
        "quantities": quantities,
        "delivered": False,
        "bill": 0
    }

    print(f"\nOrder stored for Table {table_no}: {items}\n")
    return table_no

def deliver_order(table_no):
    restaurant[table_no]["delivered"] = True
    print(f"Order delivered to Table {table_no}\n")

def generate_bill(table_no):
    items = restaurant[table_no]["order"]
    quantities = restaurant[table_no]["quantities"]

    total = 0

    for index, item in enumerate(items):
        qty = quantities[index]

        for name, price in menu.values():
            if item == name:
                total += price * qty

    restaurant[table_no]["bill"] = total
    print(f"Bill generated for Table {table_no}\n")

def print_bill(table_no):
    print("\n------ BILL ------")
    print(f"Table Number: {table_no}")
    print("Items Ordered:")

    items = restaurant[table_no]["order"]
    quantities = restaurant[table_no]["quantities"]

    for i in range(len(items)):
        print(f"{items[i]} x {quantities[i]}")

    print("Delivered:", restaurant[table_no]["delivered"])
    print("Total Bill: ₹", restaurant[table_no]["bill"])
    print("------------------\n")

table = take_order()
deliver_order(table)
generate_bill(table)
print_bill(table)
```
